# Interactive Dashboard & Tableau Storytellig


Q1. explain the difference between a tableau dashboard and a tableau story point. Give two healthcare-related use cases?

A=>
| Tableau Dashboard                                                        | Tableau Story Point                                              |
| ------------------------------------------------------------------------ | ---------------------------------------------------------------- |
| A dashboard combines multiple worksheets into a single interactive view. | A story point presents insights in a sequence like slides.       |
| Used for monitoring KPIs and exploring data.                             | Used for explaining findings and guiding users through insights. |
| Users can interact with filters and charts.                              | Users move through story points to understand the narrative.     |

**Healthcare Use Cases**
**Dashboard**:

* Monitor patient admissions, treatment costs, and department performance.

* Track hospital KPIs in real time.

**Story Point**:

* Show patient distribution by region.

* Explain treatment cost trends and outcome analysis step by step.

**How Filter Action Helps Hospital Management**

When a user selects a hospital region (e.g., North), Tableau automatically filters:

* Total patients
* Treatment costs
* Department performance
* Patient outcomes

This helps management analyze specific regions without creating separate reports.

3. What business decisions can hospital management take from these insights?

* Resource Allocation: Since Neurology and Cardiology generate the highest costs, management should review whether these departments have adequate funding, equipment, and staff, or if there are cost-saving measures that can be implemented without compromising care.

* Investigate Regional Disparities: With the South and West regions experiencing higher numbers of critical patients, management should investigate the root causes. This might require transferring specialized doctors to these regions, upgrading their intensive care units (ICUs), or launching preventative health campaigns in those specific communities.

* Doctor Training & Workload: By analyzing the Doctor Performance chart, management can identify doctors with exceptionally high patient volumes or longer-than-average patient stays, allowing them to balance workloads, prevent burnout, or provide targeted training.
